```python
import pandas as pd
import numpy as np
```

```python
prices = prices.drop(columns=['lessee_name', 'lessee_company', 'lessee_DBA'])
prices.shape
```

```python
prices = prices.dropna(subset=['cost_line_amt_USD'])
```

```python
prices['date_effective'] = pd.to_datetime(prices['date_effective'])

print(prices['date_effective'].max())
print(prices['date_effective'].min())
```

```python
prices['date_sched_termination'] = pd.to_datetime(prices['date_sched_termination'])

print(prices['date_sched_termination'].max())
print(prices['date_sched_termination'].min())
```
