## Research Question

Is there a relationship between recent street/pothole repair requests and rising city property costs (gentrification) in San Diego?
Can we predict the frequency of pothole repair requests (and the amount of time it takes to close the ticket) using the cost of city property, traffic volumes, and weather patterns in San Diego from the past decade?


## Background and Prior Work

The prevalence of potholes in San Diego is a topic of frustration for many, if not most San Diegan drivers. Within the 2,800 miles of streets and roads in San Diego, there are over 40,000 potholes that are repaired every year. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1). In San Diego, potholes most commonly arise when water seeps under pavement and weakens the integrity of the road. With the constant passage of traffic, the vibrations of vehicles driving over these weakened spots eventually break the pavement resulting in a new pothole. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). Thus, in order to reasonably predict the occurrence of potholes and the repair requests they garner, an understanding of the relationship between request frequency, weather patterns, and traffic volume must be considered. In addition, the resolution time of repair tickets may be predicted by patterns of economic growth of certain cities as well as prolonged wet periods in San Diego.


Most of the previous work related to potholes relate to the detection of potholes and not the prediction of onset/complaint. Importantly, many of the works agree that weather and traffic patterns contribute to the risk of pothole creation. One such paper focuses its research on using machine learning and in-vehicle cameras to capture 2D images of potholes and their respective size <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Related to potholes, machine learning may also be used to predict Concrete Pavement Condition Index (CPCI). A study based in South Korea seeks for more effective road management by developing a way to accurately predict the rate of degradation of concrete pavement <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). Perhaps the most similar study based its work in pothole prediction based on pavement condition indicators. This article found that the formation of potholes are mostly unpredictable and due to a multitude of conditions and dominant defects such as surface cracking and the condition of the surrounding pavement <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5).


Though these previous studies overall differ from our focus, they are nevertheless critical to our understanding of the risks of potholes as well as the importance of predictive action. They also offer more solution based literature that may only improve with the continuing development of machine learning. These works also highlight the intricacies of predicting infrastructure conditions. Thus, we acknowledge that our study is bound to many limitations and confounding factors as well as having an insufficient scale to consider many of the possible contributing factors to our question. For instance, given our present data, our focus is on pothole repair requests and not objective pothole frequency. Using repair requests as a proxy for pothole frequency has limitations in that it is not telling of pothole onset. Repeating requests may also be present. If provided a longer time-frame and sufficient resources, our current project may aim to consider more contributors to pothole formation as well as repair promptness.

## Hypothesis


We predict a strong correlation between potholes and rising city property costs in San Diego. With higher traffic volumes and increased rainfall, there is a greater frequency of pothole requests, while neighborhoods with higher property/housing values experience quicker to close repair tickets.

## Data

- The ideal variables for the pothole portion of our investigation would be where the pothole occurred (address), the time at which it occurred, the status of the request, and how many days the case existed before it closed.
- The ideal variables for the city property price portion of our research are where the property is located (address), the type of property, and the price of the leased property.
- The ideal variables for the traffic volumes portion are the number of cars in the volume, and when the observation was made.
- The ideal variables for the weather patterns part of our project are the location of the conditions (address or general region), the conditions (such as rainy, sunny, etc.), the temperature, and when the weather occurred. 

We aim for around 500 observations over the span of around the past decade from 2017-present.

We would rely on existing datasets revolving around these different aspects (potholes, city property prices, traffic volumes, and weather patterns) from the City of San Diego website, Kaggle, or other research sources. 

We would already have this data in csv spreadsheets, but will clean and organize it in our own Google sheet. 

Potholes:
- Our potholes dataset can be found at https://data.sandiego.gov/datasets/gid-pothole/. This dataset is open to the public and can have its csv file downloaded without extra steps. Variables we would use include date_requested, case_age_days to know how long it took before it closed, date_closed, status to know if it’s closed, street_address to know the location, zipcode, and public_description to understand the severity of the pothole. 

City property prices: 
- Our city property leasing prices dataset can be found at https://data.sandiego.gov/datasets/city-owned-properties-leases/. This dataset is open to the public and can have its csv file downloaded without extra steps. Variables we woud use include address_zip to know the location, lease_record_type to understand the type of property, lease_description to understand its use and role in the city, lease_status to know if it’s active, date_effective and date_sched_termination to understand when it was active, and cost_line_amt_USD to know how much it is. 

Traffic volumes:
- Our traffic volume dataset can be found at  https://data.sandiego.gov/datasets/traffic-volumes/. This dataset is open to the public and can have its csv file downloaded without extra steps. Variables we’d use include street_name to know where the observation was made, northbound_count to know the var volume going north, southbound_count  to know the var volume going south, westbound_count  to know the var volume going west, eastbound_count  to know the var volume going east, total_count to know the overall volume, and date_count to know when the observation was made. 

Weather patterns:
- Our weather patterns dataset can be found at https://oceaninformatics.ucsd.edu/datazoo/catalogs/ccelter/datasets/13. This dataset is open to the public and can have its csv file downloaded after creating an account. The variables we’d use would be date to know when it occurred, min temp to know how cold it was, max temp to know how hot it was, avg temp to get an overview of the heat for the entire day, and precipitation to know if it rained.
